```python
# 學生分數總分為 300分
# 有些學生可以加 5%

while True:
    # 這裡的程式碼會無限循環
    # 必須在某處使用 break 跳出
    try:
        scores = int(input('請輸入學生的分數(最高300分)'))
        break
    except ValueError:
        print('輸入錯誤，請輸入整數')

is_add = input('學生是否符合加分條件(y/n)?')

if is_add == "y":
    scores = scores * 1.05

#四捨五入(5進位)
scores = int(scores + 0.5)

if scores > 300:
    scores = 300

print('學生的分數是: ', scores)
```
