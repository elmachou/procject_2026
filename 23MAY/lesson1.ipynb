```python
#while 迴圈

i = 1

while i <= 10:
    print(i)
    i += 1
```

```python
# while + if example 1
i = 1

while i <= 10:
    i += 1
    if i % 2 == 0:
        continue # continue 會跳過本次迴圈剩下的程式碼，直接進入下一次while迴圈
    print(i)
    
```

```python
#while + if example 2
i = 0

while i < 10:
    i += 1

    if i == 3:
        continue # continue 會跳過本次迴圈剩下的程式碼，直接進入下一次while迴圈

    if i == 8:
        break # break 會跳出整個迴圈，不再執行迴圈內的程式碼
    
    print(i)
```

```python
#while + if example 3
while True:
    test = input("請輸入一個文字 (輸入q結束):")

    if test == "q":
        break # break 會跳出整個迴圈，不再執行迴圈內的程式碼   

    if test == "": # 如果輸入的是空字串
        continue # continue 會跳過本次迴圈剩下的程式碼，直接進入下一次while迴圈
    
    print("你輸入的文字是:", test)
```

```python
#while + if example 4
# 學生總分為300
#有些學生可以加分5%

while True:
    try:
        scores = int(input("請輸入學生分數(最高300分):"))
        break
    except ValueError:
        print("輸入錯誤，請輸入整數！")

is_add = input("學生是否符合加分條件?(y,n)")

if is_add == "y": #單項選擇
	scores = scores * 1.05

# 四捨五入(5進位)
scores = int(scores + 0.5)

# 超過300分，上限取300
if scores > 300:
	scores = 300

print("學生分數:",scores)
```

```python
# for 迴圈
for i in range(11): # range(11) 會產生從0到10的整數序列，包含0但不包含11
    print(i)

total = 0
for i in range(1, 11): # range(1, 11) 會產生從1到10的整數序列，包含1但不包含11
    total += i
print(total)
```
