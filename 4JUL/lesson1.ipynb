```python
#建立工廠
class Person:
    #建立一個實體的attribute
    def __init__(self, id):
        self.name = id

    #建立一個實體的method
    def intro(self):
        print(f"哈囉，我的名字是{self.name}")
        
```

```python
p2 = Person("robert")
p2.intro
```

```python
p3 = Person("jenny")
p3.intro
```
