```python
import yfinance as yf

ticker = "2330.TW"  # 台積電

df = yf.download(
    ticker,
    start="2020-01-01",
    end="2024-12-31",
    interval="1d",
    auto_adjust=False
)

df.tail()
```

```python
df = yf.download(
    "2330.TW",
    period="5d",
    interval="1d",
    auto_adjust=False
)
df
```

```python
close_df = df['Close'].reset_index()
close_df
```
